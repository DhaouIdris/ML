```python
import os
import sys
import json
import time
import socket
import shutil
import subprocess
from pathlib import Path
from pprint import pprint
```

```python
PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

os.chdir(PROJECT_ROOT)

def run_cmd(command, timeout=120):
    print(f"$ {command}")
    result = subprocess.run(
        command,
        shell=True,
        capture_output=True,
        text=True,
        timeout=timeout
    )
    print(result.stdout)
    if result.stderr:
        print("STDERR:")
        print(result.stderr)
    return result

def check_port(host="127.0.0.1", port=11434):
    s = socket.socket()
    s.settimeout(2)
    try:
        s.connect((host, port))
        s.close()
        return True
    except Exception:
        return False

def section(title):
    print("\n" + "=" * 80)
    print(title)
    print("=" * 80)
```

```python
section("1. Vérification de la structure projet")

required_paths = [
    "app",
    "app/main.py",
    "app/core",
    "app/agents",
    "tests",
    "data/documents",
]

missing = [p for p in required_paths if not Path(p).exists()]
if missing:
    print("Éléments manquants :", missing)
else:
    print("Structure projet OK")
```

```python
section("2. Vérification des dépendances Python")

packages = [
    "streamlit",
    "langchain",
    "langchain_community",
    "langchain_huggingface",
    "faiss",
    "pypdf",
    "pytest"
]

results = {}
for pkg in packages:
    try:
        __import__(pkg)
        results[pkg] = "OK"
    except Exception as e:
        results[pkg] = f"KO: {e}"

pprint(results)
```

```python
section("3. Vérification Ollama")

ollama_installed = shutil.which("ollama") is not None
print("Ollama installé :", ollama_installed)

if ollama_installed:
    run_cmd("ollama --version", timeout=30)
    print("Port 11434 ouvert :", check_port(port=11434))
else:
    print("Installe Ollama avant de continuer.")
```
